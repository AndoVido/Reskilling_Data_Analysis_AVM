```python
import os
import pandas as pd

def cargar_atp_csvs_por_anio(carpeta):
    """
    Carga los archivos CSV de partidos ATP desde 2000 hasta 2024.
    """
    df_matches = {}
    for anio in range(2000, 2025):
        nombre_archivo = f"wta_matches_{anio}.csv"
        ruta = os.path.join(carpeta, nombre_archivo)
        if os.path.exists(ruta):
            df = pd.read_csv(ruta)
            df_matches[anio] = df
             
        else:
            print(f" No encontrado: {nombre_archivo}")
    return df_matches
```

```python
carpeta = r"C:\Users\Usuario\Desktop\IT Academy\Tennis\CSV\Wta Matches"  # Por ejemplo: './data/atp_matches'
df_matches = cargar_atp_csvs_por_anio(carpeta)

# Ver qué años se cargaron
print(df_matches.keys())

```

```python
def unir_csvs_atp(carpeta):
    dfs = []
    for anio in range(2000, 2025):
        nombre = f"wta_matches_{anio}.csv"
        ruta = os.path.join(carpeta, nombre)
        if os.path.exists(ruta):
            df = pd.read_csv(ruta)
            df["year"] = anio  # Añade el año como columna
            dfs.append(df)
    df_final = pd.concat(dfs, ignore_index=True)
    return df_final
```

```python
df_partidos = unir_csvs_atp(r"C:\Users\Usuario\Desktop\IT Academy\Tennis\CSV\Wta Matches")
df_partidos.to_csv('partidos_wta_2000_2024.csv', index=False)

```

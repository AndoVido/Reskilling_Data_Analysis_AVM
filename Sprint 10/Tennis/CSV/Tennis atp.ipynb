PLAYERS CLEAN AND JOIN ROUNDS

```python
import pandas as pd
```

```python
players = pd.read_csv('atp_players.csv')
```

```python
players['full_name'] = players['name_first'] + " " + players['name_last']
players
```

```python
partidos = pd.read_csv('partidos_atp_2000_2024.csv')
partidos
```

```python
df_join = pd.merge(players,partidos,how='inner',left_on='player_id',right_on='winner_id')
df_join
```

```python
df_players_clean_Round = df_join.iloc[:,[0,8,5,3,6,10,14,23,58,34]]
df_players_clean_Round
```

```python
df_players_clean_Round['tourney_name'] = df_players_clean_Round['tourney_name'].str.strip().str.lower()
```

```python
df_players_clean_Round['tourney key'] = df_players_clean_Round['tourney_name'] + " - " + df_players_clean_Round['year'].astype(str)
```

```python
df_players_clean_Round
```

```python
df_players_ATP_cleaned = df_players_clean_Round[(df_players_clean_Round['tourney_name'] == 'wimbledon')  
                                        + (df_players_clean_Round['tourney_name'] == 'us open')  
                                        + (df_players_clean_Round['tourney_name'] == 'australian open')  
                                        + (df_players_clean_Round['tourney_name'] == 'roland garros')]
df_players_ATP_cleaned
```

```python
torneo_id_map = {
    'australian open': 1,
    'roland garros': 2,
    'wimbledon': 3,
    'us open': 4
}

df_players_ATP_cleaned['torneo_id'] = df_players_ATP_cleaned['tourney_name'].map(torneo_id_map)
```

```python
df_players_ATP_cleaned
```

```python
us = df_players_ATP_cleaned[df_players_ATP_cleaned['tourney_name'] == 'us open']
us
```

```python
alcar = df_players_ATP_cleaned[df_players_ATP_cleaned['full_name'] == 'Carlos Alcaraz']
alcar
```

```python
df_players_ATP_cleaned.to_csv('df_players_ATP_cleaned.csv', index=False)
```

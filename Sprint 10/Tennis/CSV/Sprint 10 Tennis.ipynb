FUNCION PARA AUTOMATIZAR LA CREACION DE COLUMNAS PARA LOS STATS POR JUGADOR

```python
def stats_totales_auto(df, group_cols=['tourney_name', 'round','year']):
    import pandas as pd
    
    # Agrupar y sumar numéricas
    grouped = df.groupby(group_cols).sum(numeric_only=True)
    
    # Detectar columnas w_ y l_
    w_cols = [c for c in grouped.columns if c.startswith('w_')]
    l_cols = [c for c in grouped.columns if c.startswith('l_')]
    
    w_metrics = {c[2:]: c for c in w_cols}
    l_metrics = {c[2:]: c for c in l_cols}
    
    # Crear totales
    for metric in w_metrics.keys() & l_metrics.keys():
        grouped[f'total_{metric}'] = grouped[w_metrics[metric]] + grouped[l_metrics[metric]]
    
    # Diccionario de porcentajes automáticos (métrica_base : métrica_referencia)
    ratios = {
        '%aces_por_punto': ('total_ace', 'total_svpt'),
        '%doble_falta': ('total_df', 'total_svpt'),
        '%primer_serve_in': ('total_1stIn', 'total_svpt'),
        '%primer_serve_won': ('total_1stWon', 'total_1stIn'),
        '%segundo_serve_won': ('total_2ndWon', 'total_svpt - total_1stIn'),
        '%bp_saved': ('total_bpSaved', 'total_bpFaced'),
        '%bp_convertidos': ('total_bpSaved', 'total_bpFaced'),  # puedes ajustarlo si quieres bp ganados
        '%juegos_de_saques_ganados': ('total_SvGms', 'total_SvGms')  # placeholder, se puede cambiar
    }
    
    # Calcular porcentajes
    for col_name, (num, den) in ratios.items():
        try:
            # Si el denominador es una resta, evaluamos
            if '-' in den:
                partes = den.split('-')
                den_val = grouped[partes[0].strip()] - grouped[partes[1].strip()]
            else:
                den_val = grouped[den]
            
            grouped[col_name] = (grouped[num] / den_val) * 100
        except KeyError:
            pass  # Si falta alguna métrica, la saltamos
    
    return grouped
```

ATTENDANCE AND TV VIEWERS

```python
import pandas as pd
import numpy as np

# años de análisis
years = list(range(2000, 2025))

# torneos base
tournaments = ["Australian Open", "Roland Garros", "Wimbledon", "US Open"]

# diccionario auxiliar con info de cada torneo
tournament_info = {
    "Australian Open": {"region": "Oceania", "surface": "Hard", "capacity": 900000},
    "Roland Garros": {"region": "Europe", "surface": "Clay", "capacity": 650000},
    "Wimbledon": {"region": "Europe", "surface": "Grass", "capacity": 500000},
    "US Open": {"region": "North America", "surface": "Hard", "capacity": 850000}
}

# crear dataset sintético
data = []
for year in years:
    for t in tournaments:
        capacity = tournament_info[t]["capacity"]

        # asistencia realista con algo de variabilidad
        attendance = np.random.randint(int(capacity*0.7), int(capacity*1.05))

        # distribución demográfica estimada
        pct_under_30 = np.random.uniform(25, 40)
        pct_over_50 = np.random.uniform(20, 35)
        pct_female = np.random.uniform(40, 50)
        pct_male = 100 - pct_female

        # métricas de TV y sesiones nocturnas
        tv_viewers_millions = np.random.uniform(40, 120)
        night_sessions = np.random.randint(5, 15)

        # cálculo de utilización
        utilization_rate = attendance / capacity

        data.append({
            "tournament": t,
            "year": year,
            "attendance": attendance,
            "capacity": capacity,
            "utilization_rate": utilization_rate,
            "pct_under_30": pct_under_30,
            "pct_over_50": pct_over_50,
            "pct_female": pct_female,
            "pct_male": pct_male,
            "region": tournament_info[t]["region"],
            "surface": tournament_info[t]["surface"],
            "tv_viewers_millions": tv_viewers_millions,
            "night_sessions": night_sessions
        })

# convertir a dataframe
df = pd.DataFrame(data)

# guardar csv final
df.to_csv("grand_slams_attendance_final_2000_2024_edit.csv", index=False)

print(df.head())

```

FUNCTION STATS JUGADORES

```python
def stats_totales_auto(df, jugadores, path_salida="./", plantilla_cols=None, referencia_csv=None):
    import os
    import numpy as np
    import pandas as pd

    # cargar columnas-plantilla para respetar nombre y orden (sinner)
    if plantilla_cols is None and referencia_csv is not None:
        plantilla_cols = list(pd.read_csv(referencia_csv, nrows=1).columns)

    # si no hay plantilla, usamos un orden razonable similar al de sinner
    if plantilla_cols is None:
        plantilla_cols = [
            'tourney_name','round','year',
            'total_2ndWon','total_1stIn','total_bpFaced',
            'total_1stWon','total_SvGms','total_df',
            'total_bpSaved','total_ace',
            '%aces_por_punto','%doble_falta',
            '%primer_serve_in','%primer_serve_won','%segundo_serve_won',
            '%bp_saved','%bp_convertidos','%juegos_de_saques_ganados'
        ]

    # eliminar el duplicado de la plantilla si aparece
    plantilla_cols = [c for c in plantilla_cols if c != 'total Ace']

    # extras que agregaremos SIN alterar el orden de la plantilla (van al final)
    extras_a_agregar = [
        '%puntos_ganados_con_servicio',  # nueva métrica
        'total_svpt',                     # serve points total (nuevo si falta)
        'w_ace', 'l_ace'                  # preservar ace por lado
    ]

    os.makedirs(path_salida, exist_ok=True)

    for jugador in jugadores['full_name']:
        df_j = df[(df['winner_name'] == jugador) | (df['loser_name'] == jugador)].copy()
        if df_j.empty:
            continue

        # agrupar
        g = df_j.groupby(['tourney_name','round','year']).sum(numeric_only=True).reset_index()

        # detectar métricas w_ / l_
        w_cols = [c for c in g.columns if c.startswith('w_')]
        l_cols = [c for c in g.columns if c.startswith('l_')]
        w_metrics = {c[2:]: c for c in w_cols}
        l_metrics = {c[2:]: c for c in l_cols}

        # crear totales (w_ + l_)
        for m in (w_metrics.keys() & l_metrics.keys()):
            g[f'total_{m}'] = g[w_metrics[m]] + g[l_metrics[m]]

        # asegurar total_ace y total_svpt
        if 'total_ace' not in g.columns and 'ace' in w_metrics and 'ace' in l_metrics:
            g['total_ace'] = g[w_metrics['ace']] + g[l_metrics['ace']]
        if 'total_svpt' not in g.columns and 'svpt' in w_metrics and 'svpt' in l_metrics:
            g['total_svpt'] = g[w_metrics['svpt']] + g[l_metrics['svpt']]

        # ratios
        def _den(df, expr):
            if isinstance(expr, str) and '-' in expr:
                a, b = [x.strip() for x in expr.split('-')]
                return df[a] - df[b]
            return df[expr]

        ratios = {
            '%aces_por_punto': ('total_ace', 'total_svpt'),
            '%doble_falta': ('total_df', 'total_svpt'),
            '%primer_serve_in': ('total_1stIn', 'total_svpt'),
            '%primer_serve_won': ('total_1stWon', 'total_1stIn'),
            '%segundo_serve_won': ('total_2ndWon', 'total_svpt - total_1stIn'),
            '%bp_saved': ('total_bpSaved', 'total_bpFaced'),
            '%bp_convertidos': ('total_bpSaved', 'total_bpFaced'),   # igual que tu sinner
            '%juegos_de_saques_ganados': ('total_SvGms', 'total_SvGms')  # mantiene 100 como en sinner
        }

        for col_name, (num, den) in ratios.items():
            if num in g.columns and (den.split('-')[0].strip() if '-' in den else den) in g.columns:
                dv = _den(g, den)
                g[col_name] = np.where(dv != 0, (g[num] / dv) * 100, np.nan)

        # nueva métrica: % puntos ganados con el servicio
        if {'total_svpt','total_1stWon','total_2ndWon'}.issubset(g.columns):
            g['%puntos_ganados_con_servicio'] = np.where(
                g['total_svpt'] != 0,
                (g['total_1stWon'] + g['total_2ndWon']) / g['total_svpt'] * 100,
                np.nan
            )

        # quitar duplicado si existe
        if 'total Ace' in g.columns and 'total_ace' in g.columns:
            g = g.drop(columns=['total Ace'])

        # construir orden final: plantilla (solo las que existan) + extras presentes
        orden_final = [c for c in plantilla_cols if c in g.columns]
        for extra in extras_a_agregar:
            if extra in g.columns and extra not in orden_final:
                orden_final.append(extra)

        # no borrar w_ace ni l_ace (ya incluidos arriba si existen)
        # exportar
        out = g[orden_final].copy()
        filename = f"{path_salida}/{jugador.replace(' ', '_')}_stats.csv"
        out.to_csv(filename, index=False)
        print(f"guardado: {filename}")
```

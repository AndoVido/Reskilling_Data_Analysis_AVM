```python
import pandas as pd
import mysql.connector
from pandas.api.types import(is_numeric_dtype,is_object_dtype,is_datetime64_any_dtype)
import matplotlib.pyplot as plt
import seaborn as sns


connect = mysql.connector.connect(
    host = 'localhost',
    user ='root',
    password = '4774',
    database = 'classicmodels'
)

cursor = connect.cursor()
cursor.execute('SHOW TABLES')
tablas_name = [row[0] for row in cursor.fetchall()]

db_classic_models = {}

for tablas in tablas_name:
    cursor.execute(f'select * from {tablas}')
    datos = cursor.fetchall()
    cursor_column_names = cursor.column_names
    df = pd.DataFrame(datos, columns = cursor_column_names)
    db_classic_models[tablas] = df

connect.close()
```

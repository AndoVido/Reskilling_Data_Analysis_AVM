-> Índice de masa corporal IMC

La función debe clasificar el resultado en sus respectivas categorías

### <span style="color: rgb(230, 215, 0);">___________________________________________________________________________________________________________________________</span>

```python
# // definir funciones //

# pedir datos al usuario
def datos_usuario():
    peso = float(input("Ingresa tu peso en kilogramos"))
    altura = float(input("Ingresa la altura en centimetros"))
    return peso,altura

# clasificar el IMC según los valores de obesidad
def composicion_corporal(clas):
    valor_peso = round(clas,2)
    if valor_peso < 18.5:
        return "peso inferior al normal, intenta consumir más proteinas y carbohidratos"
    elif valor_peso < 24.9:
        return "normal, ¡Enhorabuena!"
    elif valor_peso < 29.9:
        return "peso superior al normal, estate atento y mejora tu dieta"
    else:
        return "obesidad, ojo que estos valores preocupan mucho. ¡Cuidate!"

# mostrar clasificacion segun el resultado
def vista_clasificacion(valorpesoC,clasificado):
    print(f"Tu índice de masa corporal es de: {round(valorpesoC,2)}, lo que se considera {clasificado}")

# imprimir los resultados de la clasificacion
def resultados_clasificacion(valueIMC):
    value_clasificado = composicion_corporal(valueIMC)
    vista_clasificacion(valueIMC,value_clasificado)

# programa calculadora IMC
def programa_calculadoraIMC():
    """
    Entrada:
        Introduces peso y altura para calcular el Índice de masa corporal
    Retorno:
        El valor calculado con un mensaje con la clasificación que ese valor representa 
    Ejemplo:
    >>> Tu índice de masa corporal es de <IMC= peso / ((altura/100)**2)> lo que se considera <Composición Corporal clasificada>
    """
    peso,altura = datos_usuario()
    imc = peso / (altura/100) ** 2 #dividir altura entre 100 para convertir el valor de centimetros a metros
    resultados_clasificacion(imc)
```

Existen diversas unidades de temperatura utilizadas en distintos contextos y regiones. Las más comunes son Celsius (°C), Fahrenheit (°F) y Kelvin (K). También existen otras unidades como Rankine (°Ra) y Réaumur (°Re). Selecciona al menos 2 conversores, de modo que al introducir una temperatura devuelva, como mínimo, dos conversiones.

### <span style="color: rgb(230, 215, 0);">___________________________________________________________________________________________________________________________</span>

```python
# pedir datos al usario para convertir
def datos_temperaturas():
    temperatura = float(input("Introduce el valor que quieres convertir"))
    unidad_temperatura = input("ingresa C para <Celsius>, F para <Farenheit> o K para <Kelvin>").upper()
    return temperatura,unidad_temperatura

# creamos la funcion con las formulas de conversion

def formulas_conversor(temperatura,unidad):
    if unidad == "C": # Convierte de Celsius a F y K
        fahrenheit = temperatura * 9/5 +32
        kelvin = temperatura + 273.15
        return {
                "Fahrenheit": fahrenheit,
                "Kelvin": kelvin
                }
    elif unidad == "F": # Coniverte de Farenheit a C y K
        celsius = (temperatura - 32) * 5/9
        kelvin = (temperatura -32) * 5/9 + 273.15
        return {
                "Celsius": celsius,
                "Kelvin": kelvin
                }       
    elif unidad == "K": # Convierte de Kelvin a C y F
        celsius = temperatura - 273.15
        fahrenheit = (temperatura - 273.15) * 9/5+32
        return {
                "Celsius": celsius,
                "Fahrenheit": fahrenheit
                }
    else:
        return "se ha producido un error"

# definimos el conversor con los datos de temperatura entregado por el usuario y las formulas
def conversor():
    """
    Entrada:
        introduce el valor de la temperatura y la unidad que se desea convertir
    Retorno:
        diccionario con los valores convertidos a dos temperaturas
    """
    temperatura, unidad = datos_temperaturas()
    diccionario_resultados = formulas_conversor(temperatura, unidad)

    # bucle para iterar sobre llave y valor en el diccionario
    for llave, valor in diccionario_resultados.items():
        print(f"{llave}: {round(valor, 2)}")
```

El cliente quedó contento con el contador de palabras, pero ahora quiere leer archivos TXT y que calcule la frecuencia de cada palabra ordenadas dentro de las entradas habituales del diccionario según la letra con la que comienzan, es decir, las claves deben ir de la A a Z y dentro de la A debemos ir de la A a Z. "tu_me_quieres_blanca.txt" la salida esperada sería:

### <span style="color: rgb(230, 215, 0);">___________________________________________________________________________________________________________________________</span>

```python
import re
import string
def palabras_por_letra(ruta_archivo):
    diccionario_recuento = {}
    with open(ruta_archivo,"r",encoding="utf-8") as archivo:
        for linea in archivo:
            linea = linea.strip()
            if linea:
                palabras = linea.split(None)
                for palabra in palabras:
                    palabra = re.sub(r'[^\wáéíóúüñÁÉÍÓÚÜÑ]', '', palabra.lower())
                    if palabra:
                        letra_inicial = palabra[0].lower()

                        if letra_inicial not in diccionario_recuento:
                            diccionario_recuento[letra_inicial] = {}

                        if palabra in diccionario_recuento[letra_inicial]:
                            diccionario_recuento[letra_inicial][palabra] += 1
                        else:
                            diccionario_recuento[letra_inicial][palabra] = 1
    diccionario_ordenado = {}
    for letra in sorted(diccionario_recuento.keys()):
        diccionario_ordenado[letra] = dict(sorted(diccionario_recuento[letra].items()))

    return diccionario_ordenado
```

### <span style="color: rgb(230, 0, 126);"> - Ejercicio 1</span>  Generador de Contrasenyes

Explora el funcionament del mòdul random de la llibreria numpy. 🔗 Random module in NumPy

En aquest punt, el client ha detectat un problema amb les contrasenyes que utilitzen els seus treballadors. Asdf1234, dates d’aniversari o similars. Per a solucionar-ho ens ha encarregat una funció de Python que generi contrasenyes més segures. La funció ha de dependre dels següents paràmetres:

longitud (int): Longitud de la contrasenya
majuscules (bool = True): Si hi ha d’aparèixer majúscules
minuscules (bool = True): Si hi ha d’aparèixer minúscules
numeros (bool = True): Si hi ha d’aparèixer números
signes (bool = False): Si hi ha d’aparèixer caràcters especials (,-$? o similars)
Així doncs, si executem la funció de la següent manera:

crear_contrasenya(10, True, True, True, True)

Hauríem d’obtenir un output (que hem de poder guardar) de l’estil:

9Er,5Vn8P$

Assegura’t que es compleixin tots els criteris, i que aquestes contrasenyes siguin realment aleatòries.

(EXTRA) Explora com podríem fer que la funció copiés la contrasenya automàticament al porta papers de l’ordinador (com si l’haguéssim seleccionat i fet ctrl+copy).

### <span style="color: rgb(230, 215, 0);">___________________________________________________________________________________________________________________________</span>

```python
import numpy
import string
import pyperclip

# Generador de contraseñas aleatorias
def generador_contrasenyes(longitud = 15, usar_mayusculas = True, usar_numeros = True, usar_simbolos = False):
    """
    Genera una contraseña aleatoria de la longitud especificada.
    Parámetros:
        longitud (int): Longitud de la contraseña a generar. Por defecto es 15.
        usar_mayusculas (bool): Indica si se deben incluir letras mayúsculas. Por defecto es True.
        usar_numeros (bool): Indica si se deben incluir números. Por defecto es True.
        usar_simbolos (bool): Indica si se deben incluir símbolos. Por defecto es False.
    salida:
        La contraseña generada.
    """
    char = list(string.ascii_lowercase)

    # lista de las contraseñas obligatorias y comprobaciones
    lista_obligatoria = []
    if longitud < 4:
        raise ValueError("la longitud mínima de la contraseña debe ser de 4 caracteres")
    if usar_simbolos:
        char += list(string.punctuation)
        lista_obligatoria.append(numpy.random.choice(list(string.punctuation)))
    if usar_mayusculas:
        char += list(string.ascii_uppercase)
        lista_obligatoria.append(numpy.random.choice(list(string.ascii_uppercase)))
    if usar_numeros:
        char += list(string.digits)
        lista_obligatoria.append(numpy.random.choice(list(string.digits)))

    # Generar la contraseña
    char = numpy.array(char)
    contrasenye = numpy.random.choice(char, longitud)
    contrasenye = numpy.array(list(contrasenye) + lista_obligatoria)   
    numpy.random.shuffle(contrasenye)
    contrasenye = list(contrasenye)

    # Asegurarse de que la contraseña contiene al menos caracteres en mayusculas 
    if usar_mayusculas:
        contrasenye[0] = contrasenye[0].upper()

    # Copiar al portapapeles
    pyperclip.copy("".join(contrasenye))
    print("La contraseña ha sido copiada al portapapeles.")
    return "".join(contrasenye) 
```

### <span style="color: rgb(230, 0, 126);"> - Ejercicio 2</span> Processament de dades simple

Una compañera de trabajo nos ha pedido un favor, aprovechando que sabe que estamos aprendiendo a programar. Tiene un histórico de partidos de fútbol catalán en un fichero, donde se almacenan los nombres de los equipos y los resultados. Necesita que procesemos los datos de forma automática, para extraer los resultados que necesita.

Utilitza el fitxer "historic_partits.txt"

Necessita un programa que retorni:
- El número total de goles que ha marcado cada equipo.
- El nombre del equipo más goleador.
- El nombre del equipo más goleado
- La clasificación global (cada victoria: 3 pts, empate 1 pts, derrota 0 pts)

### <span style="color: rgb(230, 215, 0);">___________________________________________________________________________________________________________________________</span>

```python
tupla_equipos = set()
lineas_vistas = set()
partidos_ganados = {}
partidos_perdidos = {}
partidos_empatados = {}
goles = {}
goles_contra = {}
puntos = {}
partidos_jugados = {}
tabla_resultados = {}

# abrir el archivo e iterar en la lista para guardar en los diferentes diccionarios a cada equipo (llave) con su valor
with open("historic_partits.txt","r",encoding="utf-8") as archivo_futbol:
        for line in archivo_futbol:

            # limpiar y separar los datos
            local,resultado,visitante = line.strip().split('\t') # eliminar espacios, final de linea y el tabulador anidado en el codigo original
            local_goles, visitante_goles = map(int, resultado.split('-')) # split los goles y diferenciarlos por equipo local y visitante

            partido_clave = (local, local_goles, visitante, visitante_goles)

            if partido_clave in lineas_vistas:
                  continue
            lineas_vistas.add(partido_clave)

            # crear lista(tupla con set para eliminar duplicados) de los equipos
            tupla_equipos.add(local)
            tupla_equipos.add(visitante)

            # añadir equipo si no está en el diccionario
            for equipo in [local, visitante]:
                  if equipo not in partidos_ganados:
                        partidos_ganados[equipo] = 0
                        partidos_perdidos[equipo] = 0
                        partidos_empatados[equipo] = 0
                        goles[equipo] = 0
                        goles_contra[equipo] = 0
                        puntos[equipo] = 0
                        partidos_jugados[equipo] = 0

            # si gana el equipo local
            if local_goles > visitante_goles:
                  partidos_ganados[local] += 1
                  partidos_perdidos[visitante] += 1
                  partidos_jugados[local] +=1
                  partidos_jugados[visitante] += 1
                  puntos[local] += 3

            # si gana el visitante
            elif local_goles < visitante_goles:
                  partidos_ganados[visitante] += 1
                  partidos_perdidos[local] += 1
                  partidos_jugados[local] +=1
                  partidos_jugados[visitante] += 1
                  puntos[visitante] += 3

            # si hay empate
            else:
                  partidos_empatados[local] += 1
                  partidos_empatados[visitante] += 1
                  partidos_jugados[local] +=1
                  partidos_jugados[visitante] += 1
                  puntos[local] += 1
                  puntos[visitante] += 1

            # suma de goles
            goles[local] += local_goles
            goles[visitante] += visitante_goles   

            goles_contra[local] += visitante_goles
            goles_contra[visitante] += local_goles


# imprimir resultados
print("partidos_ganados:", partidos_ganados)
print("partidos_perdidos:", partidos_perdidos)
print("partidos_empatados:", partidos_empatados)
print("goles:", goles)
print("puntos:", puntos)
```

```python
# recorrer la lista de los equipos para guardar todos los diccionarios en uno solo que se convierte en la tabla de resultados
for equipo in sorted(tupla_equipos):
    ganados = partidos_ganados[equipo]
    perdidos = partidos_perdidos[equipo]
    empatados = partidos_empatados[equipo]
    goles_totales = goles[equipo]
    goals_encajados = goles_contra[equipo]
    puntos_totales = puntos[equipo]
    partidos_jugados_equipo = partidos_jugados[equipo]

    # crear diccionario un usando equipo de la tupla como llave
    tabla_resultados[equipo] = {
        "PJ":partidos_jugados_equipo,
        "G":ganados,
        "E":empatados,
        "P":perdidos,
        "GF":goles_totales,
        "GC":goals_encajados,
        "Puntos":puntos_totales
    }
# imprimir el encabezado de la tabla
print("—" * 49)
print(f"{'Equipo':<20} {'PJ':<3} {'G':<3} {'E':<3} {'P':<3} {'GF':<3} {'GC':<3} {'Pts':<4}")
print("—" * 49)

# iterar sobre el diccionario general usando el resultado como llave y (el espacio de distancia en tre columans) como valor
for equipo, datos in sorted(tabla_resultados.items(), key=lambda x: x[1]['Puntos'],reverse=True):
    print(f"{equipo:<20} {datos['PJ']:<3} {datos['G']:<3} {datos['E']:<3} {datos['P']:<3} {datos['GF']:<4} {datos['GC']:<4} {datos['Puntos']:<4}")
print("~" * 49)
```

```python
tupla_equipos = set()
lineas_vistas = set()
partidos_ganados = {}
partidos_perdidos = {}
partidos_empatados = {}
goles = {}
goles_contra = {}
puntos = {}
partidos_jugados = {}
tabla_resultados = {}

# abrir el archivo e iterar en la lista para guardar en los diferentes diccionarios a cada equipo (llave) con su valor
with open("historic_partits.txt","r",encoding="utf-8") as archivo_futbol:
        for line in archivo_futbol:
            

            # limpiar y separar los datos
            local,resultado,visitante = line.strip().split('\t') # eliminar espacios, final de linea y el tabulador anidado en el codigo original
            local_goles, visitante_goles = map(int, resultado.split('-')) # split los goles y diferenciarlos por equipo local y visitante

            partido_clave = (local, local_goles, visitante, visitante_goles)

            if partido_clave in lineas_vistas:
                  continue
            lineas_vistas.add(partido_clave)

            # crear lista(tupla con set para eliminar duplicados) de los equipos
            tupla_equipos.add(local)
            tupla_equipos.add(visitante)

            # añadir equipo si no está en el diccionario
            for equipo in [local, visitante]:
                  if equipo not in partidos_ganados:
                        partidos_ganados[equipo] = 0
                        partidos_perdidos[equipo] = 0
                        partidos_empatados[equipo] = 0
                        goles[equipo] = 0
                        goles_contra[equipo] = 0
                        puntos[equipo] = 0
                        partidos_jugados[equipo] = 0

            # si gana el equipo local
            if local_goles > visitante_goles:
                  partidos_ganados[local] += 1
                  partidos_perdidos[visitante] += 1
                  partidos_jugados[local] +=1
                  partidos_jugados[visitante] += 1
                  puntos[local] += 3

            # si hay empate
            elif local_goles == visitante_goles:
                  partidos_empatados[local] += 1
                  partidos_empatados[visitante] += 1
                  partidos_jugados[local] +=1
                  partidos_jugados[visitante] += 1
                  puntos[local] += 1
                  puntos[visitante] += 1

            # si gana visitante
            else:
                  partidos_ganados[visitante] += 1
                  partidos_perdidos[local] += 1
                  partidos_jugados[local] +=1
                  partidos_jugados[visitante] += 1
                  puntos[visitante] += 3
                  

            # suma de goles
            goles[local] += local_goles
            goles[visitante] += visitante_goles   

            goles_contra[local] += visitante_goles
            goles_contra[visitante] += local_goles


# imprimir resultados
print("partidos_ganados:", partidos_ganados)
print("partidos_perdidos:", partidos_perdidos)
print("partidos_empatados:", partidos_empatados)
print("goles:", goles)
print("puntos:", puntos)

# recorrer la lista de los equipos para guardar todos los diccionarios en uno solo que se convierte en la tabla de resultados
for equipo in sorted(tupla_equipos):
    ganados = partidos_ganados[equipo]
    perdidos = partidos_perdidos[equipo]
    empatados = partidos_empatados[equipo]
    goles_totales = goles[equipo]
    goals_encajados = goles_contra[equipo]
    puntos_totales = puntos[equipo]
    partidos_jugados_equipo = partidos_jugados[equipo]

    # crear diccionario un usando equipo de la tupla como llave
    tabla_resultados[equipo] = {
        "PJ":partidos_jugados_equipo,
        "G":ganados,
        "E":empatados,
        "P":perdidos,
        "GF":goles_totales,
        "GC":goals_encajados,
        "Puntos":puntos_totales
    }
# imprimir el encabezado de la tabla
print("—" * 49)
print(f"{'Equipo':<20} {'PJ':<3} {'G':<3} {'E':<3} {'P':<3} {'GF':<3} {'GC':<3} {'Pts':<4}")
print("—" * 49)

# iterar sobre el diccionario general usando el resultado como llave y (el espacio de distancia en tre columans) como valor
for equipo, datos in sorted(tabla_resultados.items(), key=lambda x: x[1]['Puntos'],reverse=True):
    print(f"{equipo:<20} {datos['PJ']:<3} {datos['G']:<3} {datos['E']:<3} {datos['P']:<3} {datos['GF']:<4} {datos['GC']:<4} {datos['Puntos']:<4}")
print("~" * 49)
```

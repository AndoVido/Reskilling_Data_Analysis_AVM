```python
tupla_equipos = set()
partidos_ganados = {}
partidos_perdidos = {}
partidos_empatados = {}
goles = {}
puntos = {}
tabla_resultados = {}
partidos_jugados = {}

with open("historic_partits.txt","r",encoding="utf-8") as archivo_abrir:
    for line in archivo_abrir:
        equipo_local,resultado_partido,equipo_visitante = line.strip().split('\t')

        goles_e_local, goles_e_visitante = map(int, resultado_partido.split('-'))
        tupla_equipos.add(equipo_local)

        for equipo in [equipo_local, equipo_visitante]:
            if equipo not in partidos_ganados:
                partidos_ganados[equipo] = 0
                partidos_perdidos[equipo] = 0
                partidos_empatados[equipo] = 0
                goles[equipo] = 0
                puntos[equipo] = 0

        if goles_e_local > goles_e_visitante:
            partidos_ganados[equipo_local] += 1
            partidos_perdidos[equipo_visitante] += 1
            puntos[equipo_local] += 3
            partidos_jugados[equipo_local] += 1

        if goles_e_visitante > goles_e_local:
            partidos_ganados[equipo_visitante] += 1
            partidos_perdidos[equipo_local] += 1
            puntos[equipo_visitante] += 3
            partidos_jugados[equipo_visitante] += 1
        else:
            partidos_empatados[equipo_local] += 1
            partidos_empatados[equipo_visitante] += 1
            puntos[equipo_local] += 1
            puntos[equipo_visitante] += 1
            partidos_jugados[equipo_local] +=1
            partidos_jugados[equipo_visitante] += 1

    for equipo in sorted(tupla_equipos):
        played = partidos_jugados[equipo]
        win = partidos_ganados[equipo]
        lose = partidos_perdidos[equipo]
        tie = partidos_empatados[equipo]
        goals = goles[equipo]
        points = puntos[equipo]
        tabla_resultados[equipo] = {
            "PJ":played,
            "G":win,
            "E":tie,
            "P":lose,
            "Goles":goals,
            "Puntos":points
        }
print("—" * 44)
print(f"{'Equipo':<20} {'PJ':<3} {'G':<3} {'E':<3} {'P':<3} {'GF':<3} {'Pts':<4}")
print("—" * 44)

for equipo, datos in sorted(tabla_resultados.items(), key=lambda x: x[1]['Goles'],reverse=True):
    print(f"{equipo:<20} {datos['PJ']:<3} {datos['G']:<3} {datos['E']:<3} {datos['P']:<3} {datos['Goles']:<4} {datos['Puntos']:<4}")
print("~" * 44)
```

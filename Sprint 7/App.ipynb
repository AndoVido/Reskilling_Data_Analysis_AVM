```python
import streamlit as st # type: ignore
import numpy as np
import string
import pyperclip

# Función generadora de contraseñas
def generador_contrasenyes(longitud=15, usar_mayusculas=True, usar_numeros=True, usar_simbolos=False):
    char = list(string.ascii_lowercase)
    lista_obligatoria = []

    if longitud < 4:
        raise ValueError("La longitud mínima de la contraseña debe ser de 4 caracteres")

    if usar_simbolos:
        char += list(string.punctuation)
        lista_obligatoria.append(np.random.choice(list(string.punctuation)))
    if usar_mayusculas:
        char += list(string.ascii_uppercase)
        lista_obligatoria.append(np.random.choice(list(string.ascii_uppercase)))
    if usar_numeros:
        char += list(string.digits)
        lista_obligatoria.append(np.random.choice(list(string.digits)))

    char = np.array(char)
    contrasenye = np.random.choice(char, longitud)
    contrasenye = np.array(list(contrasenye) + lista_obligatoria)
    np.random.shuffle(contrasenye)
    contrasenye = list(contrasenye)

    if usar_mayusculas:
        contrasenye[0] = contrasenye[0].upper()

    contrasenye = "".join(contrasenye)
    pyperclip.copy(contrasenye)
    return contrasenye

# Interfaz Streamlit
st.title("🔐 Generador de Contraseñas Seguras")

longitud = st.slider("Longitud de la contraseña", 4, 50, 15)
usar_mayusculas = st.checkbox("Incluir mayúsculas", True)
usar_numeros = st.checkbox("Incluir números", True)
usar_simbolos = st.checkbox("Incluir símbolos", False)

if st.button("Generar contraseña"):
    try:
        password = generador_contrasenyes(longitud, usar_mayusculas, usar_numeros, usar_simbolos)
        st.success("Contraseña generada y copiada al portapapeles:")
        st.code(password, language="text")
    except Exception as e:
        st.error(f"Error: {e}")
```

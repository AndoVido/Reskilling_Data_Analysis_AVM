```python
# Defino la función para calcular la ruta optima y que al final se imprima como df

def calcular_ruta_optima(dataframe):
    import pandas as pd
    ciudad_actual = 'Barcelona'
    ruta = []

    while len(dataframe) > 0:
        matriz = dataframe[ciudad_actual]
        lado_B = matriz.idxmin()
        distancia = matriz.min()
        ruta.append((ciudad_actual, lado_B, distancia))
        dataframe = dataframe.drop(ciudad_actual,axis=0).drop(ciudad_actual, axis=1)
        ciudad_actual = lado_B
    df_ruta = pd.DataFrame(ruta, columns=['Punto A','Punto B','Distancia A → B']) # para filtrar '→' usa el código ASCII 'Alt + 26' 
    return df_ruta
```

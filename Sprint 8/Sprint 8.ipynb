```python
datos_fecha = df_trabajadores_index[['Dia','Mes','Any']]
datos_fecha = datos_fecha.rename(columns={'Dia':'day','Mes':'month','Any':'year'})
df_trabajadores_index['Fecha Nacimiento'] = pd.to_datetime(datos_fecha)
df_trabajadores_index
```

```python
print(datos_fecha.columns)
```

```python
hoy = pd.to_datetime('today')
df_trabajadores_index['Edad Actual'] = df_trabajadores_index['Fecha Nacimiento'].apply(lambda fecha: hoy.year - fecha.year - ((hoy.month, hoy.day) < (fecha.month, fecha.day)))
df_trabajadores_index
```

### <span style="color: rgb(230, 0, 126);"> - Ejercicio 1</span>
Utilizando el siguiente DataFrame, adjunta la columna "Incremento" al dataframe del nivel anterior.
Actualiza la columna salario en función de los porcentajes que se adjuntan. No modifiques manualmente los incrementos, escribe código Python para realizar las conversiones necesarias.

### <span style="color: rgb(230, 215, 0);">___________________________________________________________________________________________________________________________</span>

```python
df_increment = pd.DataFrame({"Grup":["Grup A","Grup B","Grup C", "Grup D" ] ,  "Increment": ["5%","3,5%","2%","8%"]})
```

```python
df_trabajadores_incremento = df_trabajadores_index.merge(df_increment, left_on='Grup Professional', right_on='Grup')
df_trabajadores_incremento
```

```python
# limpiar el simbolo de porcentaje
df_trabajadores_incremento['Increment'] = df_trabajadores_incremento['Increment'].replace('[%,.]', '', regex=True).astype(float).round(2) / 100
```

```python
df_trabajadores_incremento['Salari mensual'] = (df_trabajadores_incremento['Salari mensual'] * (1 + df_trabajadores_incremento['Increment'])).round(2)
df_trabajadores_incremento
```

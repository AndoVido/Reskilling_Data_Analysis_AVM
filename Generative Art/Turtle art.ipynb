```python
from turtle import *
import random


setup(1000,1000)

# x='x axis' , y='y axis' , s='size' , l='level of recursion(max 8 levels)'
def tiling(x,y,s,l,mode='straight'):

    if l == 0:

        if mode == 'straight':
            # vertical line
            if random.random() < 0.5:
                penup()
                goto(x,y-s)
                pendown()
                goto(x,y+s)

            else:
                # horizontal line
                penup()
                goto(x-s,y)
                pendown()
                goto(x+s,y)
        
        elif mode == 'diagonal':
            # top left to bottom right
                if random.random() < 0.5:
                    penup()
                    goto(x-s,y+s)
                    pendown()
                    goto(x+s,y-s)

                else:
                    # bottom left to top right
                    penup()
                    goto(x-s,y-s)
                    pendown()
                    goto(x+s,y+s)
    else:
        s /= 2
        l -= 1
        tiling(x-s, y+s,s,l,mode)
        tiling(x+s, y+s,s,l,mode)
        tiling(x-s, y-s,s,l,mode)
        tiling(x+s, y-s,s,l,mode)


def start():
    tiling(0, 0, 400, 5, mode='diagonal')
    tracer(True)

hideturtle()
tracer(False)
ontimer(start, 100)  # espera 100 milisegundos antes de comenzar
exitonclick()
```

```python
from turtle import *
import random

setup(1200, 1200)

def tiling(x, y, s, l, mode='straight'):
    if l == 0:
        if mode == 'straight':
            if random.random() < 0.5:
                penup(); goto(x, y - s); pendown(); goto(x, y + s)
            else:
                penup(); goto(x - s, y); pendown(); goto(x + s, y)
        elif mode == 'diagonal':
            if random.random() < 0.5:
                penup(); goto(x - s, y + s); pendown(); goto(x + s, y - s)
            else:
                penup(); goto(x - s, y - s); pendown(); goto(x + s, y + s)
    else:
        s /= 2
        l -= 1
        tiling(x - s, y + s, s, l, mode)
        tiling(x + s, y + s, s, l, mode)
        tiling(x - s, y - s, s, l, mode)
        tiling(x + s, y - s, s, l, mode)

def start():
    hideturtle()
    tracer(False)
    tiling(0, 0, 400, 5, mode='diagonal')
    tracer(True)

ontimer(start, 100)
exitonclick()
```
